```python
from collections import deque

graph = {
    "Andheri": [("WEH", 2.0), ("Santacruz", 3.0)],
    "WEH": [("Vakola", 2.5)],
    "Vakola": [("Bandra Flyover", 3.0)],
    "Bandra Flyover": [("BKC", 3.6)],
    "Santacruz": [("Kalina", 3.0)],
    "Kalina": [("BKC", 5.1)],
    "BKC": []
}

def bfs(graph, start, goal):
    q = deque()
    q.append((start, [start], 0))

    while q:
        current, route, total = q.popleft()

        if current == goal:
            return route, total

        for next_node, dist in graph[current]:
            q.append((next_node, route + [next_node], total + dist))

route, distance = bfs(graph, "Andheri", "BKC")

print("BFS Route:")
print(" -> ".join(route))
print("Total Distance:", distance, "km")
```

```python
graph = {
    "Andheri": [("WEH", 2.0), ("Santacruz", 3.0)],
    "WEH": [("Vakola", 2.5)],
    "Vakola": [("Bandra Flyover", 3.0)],
    "Bandra Flyover": [("BKC", 3.6)],
    "Santacruz": [("Kalina", 3.0)],
    "Kalina": [("BKC", 5.1)],
    "BKC": []
}

def dfs(graph, start, goal):
    stack = [(start, [start], 0)]

    while stack:
        current, route, total = stack.pop()

        if current == goal:
            return route, total

        for next_node, dist in reversed(graph[current]):
            stack.append((next_node, route + [next_node], total + dist))

route, distance = dfs(graph, "Andheri", "BKC")

print("DFS Route:")
print(" -> ".join(route))
print("Total Distance:", distance, "km")
```

```python
vertices = 7
edges = 7

bfs_path = "[1, 5, 6, 7]"
dfs_path = "[1, 2, 3, 4, 7]"

print("Comparison of BFS and DFS")
print("-" * 40)

print(f"Vertices (V) = {vertices}")
print(f"Edges (E) = {edges}\n")

print("BFS Path :", bfs_path)
print("Distance : 11.1 km")
print("Stops    : 4\n")

print("DFS Path :", dfs_path)
print("Distance : 11.1 km")
print("Stops    : 5\n")

print("Time Complexity")
print(f"BFS = O(V + E) = O({vertices} + {edges})")
print(f"DFS = O(V + E) = O({vertices} + {edges})\n")

print("Space Complexity")
print("BFS = O(V)")
print("DFS = O(V)\n")

print("Conclusion:")
print("BFS is better for navigation because it finds the shortest path.")
print("DFS explores one path completely before backtracking.")
```
